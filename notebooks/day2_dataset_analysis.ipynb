```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
feature_importance = pd.Series(
    rf.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

feature_importance.head(15)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
y_pred = rf.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Phishing", "Legitimate"],
    yticklabels=["Phishing", "Legitimate"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Random Forest Confusion Matrix")
plt.show()
```

```python
deployment_features = [
    "URLLength",
    "DomainLength",
    "IsDomainIP",
    "TLDLength",
    "NoOfSubDomain",
    "HasObfuscation",
    "NoOfObfuscatedChar",
    "ObfuscationRatio",
    "NoOfLettersInURL",
    "LetterRatioInURL",
    "NoOfDegitsInURL",
    "DegitRatioInURL",
    "NoOfEqualsInURL",
    "NoOfQMarkInURL",
    "NoOfAmpersandInURL",
    "NoOfOtherSpecialCharsInURL",
    "SpacialCharRatioInURL",
    "IsHTTPS"
]

print(len(deployment_features))
```

```python
X_deployment = df_model[deployment_features]
y_deployment = df_model["label"]

print(X_deployment.shape)
print(y_deployment.shape)
```

```python
X_dep_train, X_dep_test, y_dep_train, y_dep_test = train_test_split(
    X_deployment,
    y_deployment,
    test_size=0.2,
    random_state=42,
    stratify=y_deployment
)

print("Training:", X_dep_train.shape)
print("Testing :", X_dep_test.shape)
```

```python
xgb_deployment = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    n_jobs=-1,
    eval_metric="logloss"
)

xgb_deployment.fit(X_dep_train, y_dep_train)
```

```python
y_dep_pred = xgb_deployment.predict(X_dep_test)

print("Accuracy :", accuracy_score(y_dep_test, y_dep_pred))
print("Precision:", precision_score(y_dep_test, y_dep_pred))
print("Recall   :", recall_score(y_dep_test, y_dep_pred))
print("F1 Score :", f1_score(y_dep_test, y_dep_pred))
```
