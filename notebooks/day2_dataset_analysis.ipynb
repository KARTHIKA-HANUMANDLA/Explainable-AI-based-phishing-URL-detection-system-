```python
import pandas as pd

features_to_compare = [
    "URLLength",
    "DomainLength",
    "IsDomainIP",
    "TLDLength",
    "NoOfSubDomain",
    "HasObfuscation",
    "NoOfObfuscatedChar",
    "ObfuscationRatio",
    "NoOfLettersInURL",
    "LetterRatioInURL",
    "NoOfDegitsInURL",
    "DegitRatioInURL",
    "NoOfEqualsInURL",
    "NoOfQMarkInURL",
    "NoOfAmpersandInURL",
    "NoOfOtherSpecialCharsInURL",
    "SpacialCharRatioInURL",
    "IsHTTPS"
]

# Take 20 URLs directly from the dataset
sample_df = df[["URL"] + features_to_compare].head(20).copy()

comparison_rows = []

for _, row in sample_df.iterrows():

    extracted = extract_url_features(row["URL"])

    for feature in features_to_compare:

        actual = row[feature]
        extracted_value = extracted[feature]

        comparison_rows.append({
            "URL": row["URL"],
            "Feature": feature,
            "Dataset_Value": actual,
            "Our_Value": extracted_value,
            "Match": abs(float(actual) - float(extracted_value)) < 1e-9
        })

comparison_df = pd.DataFrame(comparison_rows)

comparison_df
```

```python
mismatches = comparison_df[comparison_df["Match"] == False]

mismatches
```

```python
mismatch_summary = (
    mismatches
    .groupby("Feature")
    .size()
    .sort_values(ascending=False)
)

mismatch_summary
```

```python
comparison_df["Absolute_Error"] = (
    comparison_df["Dataset_Value"].astype(float)
    - comparison_df["Our_Value"].astype(float)
).abs()

feature_error = (
    comparison_df
    .groupby("Feature")["Absolute_Error"]
    .mean()
    .sort_values(ascending=False)
)

feature_error
```

```python
df["our_letters"] = df["URL"].str.count(r"[A-Za-z]")

df["letter_diff"] = (
    df["our_letters"] - df["NoOfLettersInURL"]
)

for diff in [9, 5, 8, 7]:
    print(f"\n========== DIFFERENCE {diff} ==========")

    print(
        df[df["letter_diff"] == diff][
            [
                "URL",
                "URLLength",
                "NoOfLettersInURL",
                "Domain"
            ]
        ].head(5).to_string(index=False)
    )
```

```python
deployment_features = [
    "URLLength",
    "DomainLength",
    "IsDomainIP",
    "TLDLength",
    "NoOfSubDomain",
    "HasObfuscation",
    "NoOfObfuscatedChar",
    "ObfuscationRatio",
    "NoOfLettersInURL",
    "LetterRatioInURL",
    "NoOfDegitsInURL",
    "DegitRatioInURL",
    "NoOfEqualsInURL",
    "NoOfQMarkInURL",
    "NoOfAmpersandInURL",
    "NoOfOtherSpecialCharsInURL",
    "SpacialCharRatioInURL",
    "IsHTTPS"
]
```

```python
X_custom = []

for url in df["URL"]:
    features = extract_url_features(url)
    X_custom.append([
        features[f] for f in deployment_features
    ])

X_custom = pd.DataFrame(
    X_custom,
    columns=deployment_features
)

y_custom = df["label"]
```

```python
print(X_custom.shape)
print(X_custom.head())
```

```python
from sklearn.model_selection import train_test_split

X_custom_train, X_custom_test, y_custom_train, y_custom_test = train_test_split(
    X_custom,
    y_custom,
    test_size=0.2,
    random_state=42,
    stratify=y_custom
)

print("Training:", X_custom_train.shape)
print("Testing :", X_custom_test.shape)
```

```python
from xgboost import XGBClassifier

xgb_custom = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    n_jobs=-1,
    eval_metric="logloss"
)

xgb_custom.fit(X_custom_train, y_custom_train)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

y_custom_pred = xgb_custom.predict(X_custom_test)

print("Accuracy :", accuracy_score(y_custom_test, y_custom_pred))
print("Precision:", precision_score(y_custom_test, y_custom_pred))
print("Recall   :", recall_score(y_custom_test, y_custom_pred))
print("F1 Score :", f1_score(y_custom_test, y_custom_pred))
```

```python
y_custom_prob = xgb_custom.predict_proba(X_custom_test)[:, 1]

print("ROC-AUC:", roc_auc_score(y_custom_test, y_custom_prob))
```
